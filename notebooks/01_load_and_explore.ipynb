# 01 · Load real EEG by code

**Goal:** fetch a real motor-imagery dataset with *zero* manual downloads, and get a feel for what EEG data actually is.

We use MOABB's `BNCI2014_001` (BCI Competition IV-2a): 9 subjects, 22 EEG channels, motor imagery. MOABB downloads + caches it under `~/mne_data` on first call.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent / "src"))
%load_ext autoreload
%autoreload 2
import numpy as np, matplotlib.pyplot as plt
```

```python
from eeglog.data import load_moabb
# Start with 3 subjects, left-vs-right (2-class) for speed.
d = load_moabb(subjects=[1, 2, 3], paradigm='left_right')
d
```

`X` is `(n_epochs, n_channels, n_times)`. `groups` holds the subject id per epoch — that's what makes subject-independent CV possible later.

```python
print('X:', d.X.shape, '| sfreq:', d.sfreq, 'Hz')
print('classes:', sorted(set(d.y)))
# Class balance — note any imbalance now so we pick the right metric later.
import collections; print(collections.Counter(d.y))
```

```python
# Artifact: plot one trial (a few channels) and the class counts.
fig, ax = plt.subplots(1, 2, figsize=(11, 3))
t = np.arange(d.X.shape[-1]) / d.sfreq
for ch in range(5):
    ax[0].plot(t, d.X[0, ch] * 1e6 + ch * 30, lw=.6)
ax[0].set(title=f'One trial ({d.y[0]})', xlabel='s', ylabel='µV (offset)')
vals, cnts = np.unique(d.y, return_counts=True)
ax[1].bar(vals, cnts); ax[1].set_title('class balance')
plt.tight_layout()
```

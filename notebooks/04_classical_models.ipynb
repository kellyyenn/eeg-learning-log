# 04 · Classical models

Five complete pipelines (features + classifier in one estimator):
`CSP+LDA`, `FBCSP+SVM`, `Riemann+TS+LR`, `Riemann+MDM`, `BandPower+RF`.

We score them *within-subject* here for a quick read; the honest cross-subject ranking comes in notebook 06.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent / "src"))
%load_ext autoreload
%autoreload 2
import numpy as np, matplotlib.pyplot as plt
```

```python
from eeglog.data import load_moabb
from eeglog.models_classic import all_classical
from eeglog import evaluation as E
d = load_moabb(subjects=[1, 2, 3], paradigm='left_right')
models = all_classical(d.sfreq)
list(models)
```

```python
results = []
for name, model in models.items():
    r = E.evaluate_within_subject(model, d.X, d.y, d.groups, model_name=name)
    print(r); results.append(r)
```

```python
lb = E.leaderboard(results); display(lb)
plt.barh(lb['model'], lb['mean'], xerr=lb['std']); plt.xlabel('balanced acc')
plt.title('within-subject (NOT the headline)'); plt.gca().invert_yaxis()
```

**Checkpoint:** a leaderboard of classical models. Riemannian methods usually lead. Remember: these are *within-subject* (optimistic).
